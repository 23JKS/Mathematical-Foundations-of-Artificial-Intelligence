```python
from torchvision.transforms import functional as VF

from model import DWT, HAAR_G, HAAR_H, IDWT
from utility import load_image, save_image
```

```python
x = load_image(r'./data/test.jpg')
shape = [x.size(-2), x.size(-1)]
x = VF.resize(x, [4096, 4096])
dwt = DWT(h=HAAR_H, g=HAAR_G, n=12)
w = dwt.forward(x)
save_image(VF.resize(w, shape), r'./data/dwt.jpg')
```

```python
idwt = IDWT(h=HAAR_H, g=HAAR_G, n=12)
y = idwt.forward(w)
save_image(VF.resize(y, shape), r'./data/idwt.jpg')
```
